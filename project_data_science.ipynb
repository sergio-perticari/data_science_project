```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
# 1. Caricamento dei dati dai due fogli del file Excel
file_input = "online_retail_II.xlsx"
excel_data = pd.ExcelFile(file_input)
df_year1 = pd.read_excel(file_input, sheet_name=excel_data.sheet_names[0])
df_year2 = pd.read_excel(file_input, sheet_name=excel_data.sheet_names[1])
```

Sezione dedicata alla pulizia dei dati

```python

print("Valori mancanti per colonna prima della pulizia:")
print(df_year1.isnull().sum(), "\n")
"""la funzione .sum() consente di sommare i True ed i False,
sapendo che in python True vale 1 e False vale 0, il numero mostrato accanto alle
8 colonne indica quanti 'buchi' sono stati trovati
"""
def pulisci_dati(df):

    #Rimuovo duplicati
    df_clean = df.drop_duplicates()

    # Maschere booleane per filtrare quantità e prezzi maggiori di zero
    mask_quantity = df_clean['Quantity'] > 0
    mask_price = df_clean['Price'] > 0

    # Applicazione delle maschere al DataFrame
    df_clean = df_clean[mask_quantity & mask_price]

    # Rimozione dei valori mancanti
    df_clean = df_clean.dropna(subset=['Customer ID'])

    return df_clean

    # 3. Applicazione della funzione a entrambi i DataFrame
df_year1_clean = pulisci_dati(df_year1)
df_year2_clean = pulisci_dati(df_year2)



    # Mostra il conteggio esatto dei valori mancanti per ogni singola colonna del DataFrame
    # print(df_year1.isnull().sum())

    # df_clean = df_clean.dropna(subset=['Customer ID'])



print("Anno 1 - Prima:", df_year1.shape, "| Dopo:", df_year1_clean.shape)
print("Anno 2 - Prima:", df_year2.shape, "| Dopo:", df_year2_clean.shape)
print("Dimensioni Primo Foglio:", df_year1.shape)
print("Dimensioni Secondo Foglio:", df_year2.shape)
```
